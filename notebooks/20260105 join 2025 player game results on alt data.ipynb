```python
# =============================================================================
# 1. QUICK DATA OVERVIEW
# =============================================================================

print("Dataset Shape:", df.shape)
print("\nColumns:")
for i, col in enumerate(df.columns, 1):
    print(f"  {i:2d}. {col}")
    
print("\n" + "="*80)
print("Key Columns Check:")
print("="*80)
print(f"Total rows: {len(df):,}")
print(f"Rows with points_line: {df['points_line'].notna().sum():,} ({df['points_line'].notna().sum()/len(df)*100:.1f}%)")
print(f"Rows with actual PTS: {df['PTS'].notna().sum():,} ({df['PTS'].notna().sum()/len(df)*100:.1f}%)")
print(f"Rows with BOTH: {(df['points_line'].notna() & df['PTS'].notna()).sum():,}")

print(f"\nDate range: {df['game_date'].min()} to {df['game_date'].max()}")
print(f"Unique players: {df['PLAYER_NAME'].nunique():,}")
print(f"Unique games: {df['GAME_ID'].nunique():,}")
```

```python
# =============================================================================
# 2. FILTER TO ACTIONABLE DATA (has both line and actual points)
# =============================================================================

df_bets = df[(df['points_line'].notna()) & (df['PTS'].notna())].copy()

print(f"Filtered to {len(df_bets):,} actionable player-games")
print(f"  Date range: {df_bets['game_date'].min()} to {df_bets['game_date'].max()}")
print(f"  Unique players: {df_bets['PLAYER_NAME'].nunique():,}")
print(f"  Unique games: {df_bets['GAME_ID'].nunique():,}")

# Add key calculated columns
df_bets['points_diff'] = df_bets['PTS'] - df_bets['points_line']  # positive = beat line (over wins)
df_bets['over_hit'] = (df_bets['points_diff'] > 0).astype(int)    # 1 if over wins
df_bets['under_hit'] = (df_bets['points_diff'] < 0).astype(int)   # 1 if under wins
df_bets['push'] = (df_bets['points_diff'] == 0).astype(int)       # 1 if push

print(f"\n{'='*80}")
print("Overall Hit Rates:")
print(f"{'='*80}")
print(f"Overs hit:  {df_bets['over_hit'].sum():,} / {len(df_bets):,} ({df_bets['over_hit'].mean()*100:.1f}%)")
print(f"Unders hit: {df_bets['under_hit'].sum():,} / {len(df_bets):,} ({df_bets['under_hit'].mean()*100:.1f}%)")
print(f"Pushes:     {df_bets['push'].sum():,} ({df_bets['push'].mean()*100:.1f}%)")

print(f"\nAvg line: {df_bets['points_line'].mean():.1f}")
print(f"Avg actual: {df_bets['PTS'].mean():.1f}")
print(f"Avg diff: {df_bets['points_diff'].mean():.2f}")
```

```python
# =============================================================================
# 3. SAMPLE DATA - Top Props by Points Line
# =============================================================================

print("Sample: Top 10 highest lines")
sample_cols = ['PLAYER_NAME', 'game_date', 'TEAM_NAME', 'points_line', 'PTS', 'points_diff', 
               'over_hit', 'is_home', 'is_favorite', 'MIN']
print(df_bets.nlargest(10, 'points_line')[sample_cols].to_string(index=False))

print("\n" + "="*80)
print("Sample: Top 10 biggest over performances")
print(df_bets.nlargest(10, 'points_diff')[sample_cols].to_string(index=False))

print("\n" + "="*80)
print("Sample: Top 10 biggest under performances")
print(df_bets.nsmallest(10, 'points_diff')[sample_cols].to_string(index=False))
```

```python
# =============================================================================
# 7. REMOVE INJURY/EARLY EXIT GAMES (Played ≤25% of usual minutes)
# =============================================================================

print("Filtering out injury/early exit games...")
print("="*80)

# Calculate each player's average minutes across all games
player_avg_min = df_bets.groupby('PLAYER_NAME')['MIN'].mean().to_dict()
df_bets['player_avg_min'] = df_bets['PLAYER_NAME'].map(player_avg_min)

# Calculate what % of their usual minutes they played
df_bets['min_pct_of_avg'] = df_bets['MIN'] / df_bets['player_avg_min']

# Flag games where they played ≤25% of usual (likely injury/ejection/DNP-CD)
df_bets['early_exit'] = df_bets['min_pct_of_avg'] <= 0.25

print(f"Total games: {len(df_bets):,}")
print(f"Early exit games (≤25% usual minutes): {df_bets['early_exit'].sum():,} ({df_bets['early_exit'].mean()*100:.1f}%)")
print(f"\nEarly exit over hit rate: {df_bets[df_bets['early_exit']]['over_hit'].mean()*100:.1f}%")
print(f"Full game over hit rate: {df_bets[~df_bets['early_exit']]['over_hit'].mean()*100:.1f}%")

# Show examples of early exits
print("\nExample early exit games (likely injuries):")
early_exits = df_bets[df_bets['early_exit']].nlargest(10, 'points_line')
print(early_exits[['PLAYER_NAME', 'game_date', 'MIN', 'player_avg_min', 'PTS', 'points_line', 'points_diff']].to_string(index=False))

# Create clean dataset
df_clean = df_bets[~df_bets['early_exit']].copy()
print(f"\n✅ Clean dataset: {len(df_clean):,} games ({len(df_clean)/len(df_bets)*100:.1f}% of original)")
print(f"Overall over hit rate (clean data): {df_clean['over_hit'].mean()*100:.1f}%")
```

```python
# =============================================================================
# 8. MATRIX ANALYSIS: Points Line Tier × Team Spread
# =============================================================================

print("Building matrix: Player Tier (by points line) × Team Spread")
print("="*80)

# Create bins for points line (player tier)
def bin_points_line(line):
    if pd.isna(line):
        return 'Unknown'
    elif line >= 30:
        return '30+ (Superstar)'
    elif line >= 20:
        return '20-30 (Star)'
    elif line >= 10:
        return '10-20 (Role Player)'
    else:
        return '<10 (Bench)'

# Create bins for team spread
def bin_team_spread(spread):
    if pd.isna(spread):
        return 'Unknown'
    elif spread >= 10:
        return '10+ Fav'
    elif spread >= 5:
        return '5-10 Fav'
    elif spread >= 0:
        return '0-5 Fav'
    elif spread >= -5:
        return '0-5 Dog'
    elif spread >= -10:
        return '5-10 Dog'
    else:
        return '10+ Dog'

df_clean['line_tier'] = df_clean['points_line'].apply(bin_points_line)
df_clean['spread_bin'] = df_clean['team_spread'].apply(bin_team_spread)

# Create matrix: count of games
print("\n1️⃣ SAMPLE SIZE MATRIX (number of games)")
print("-" * 80)
count_matrix = pd.crosstab(
    df_clean['line_tier'], 
    df_clean['spread_bin'],
    margins=True,
    margins_name='Total'
)
# Order columns properly
col_order = ['10+ Dog', '5-10 Dog', '0-5 Dog', '0-5 Fav', '5-10 Fav', '10+ Fav', 'Total']
row_order = ['30+ (Superstar)', '20-30 (Star)', '10-20 (Role Player)', '<10 (Bench)', 'Total']
count_matrix = count_matrix.reindex(columns=[c for c in col_order if c in count_matrix.columns])
count_matrix = count_matrix.reindex([r for r in row_order if r in count_matrix.index])
print(count_matrix.to_string())

# Create matrix: over hit rate %
print("\n\n2️⃣ OVER HIT RATE MATRIX (% of overs that hit)")
print("-" * 80)
over_matrix = pd.crosstab(
    df_clean['line_tier'], 
    df_clean['spread_bin'],
    values=df_clean['over_hit'],
    aggfunc='mean'
) * 100
over_matrix = over_matrix.reindex(columns=[c for c in col_order[:-1] if c in over_matrix.columns])
over_matrix = over_matrix.reindex([r for r in row_order[:-1] if r in over_matrix.index])
print(over_matrix.round(1).to_string())

# Add overall rates for comparison
print("\n\n3️⃣ BENCHMARKS")
print("-" * 80)
print(f"Overall over hit rate: {df_clean['over_hit'].mean()*100:.1f}%")
print("\nBy Player Tier:")
for tier in row_order[:-1]:
    tier_data = df_clean[df_clean['line_tier'] == tier]
    if len(tier_data) > 0:
        print(f"  {tier:20s}: {tier_data['over_hit'].mean()*100:.1f}% ({len(tier_data):,} games)")

print("\nBy Team Spread:")
for spread in col_order[:-1]:
    spread_data = df_clean[df_clean['spread_bin'] == spread]
    if len(spread_data) > 0:
        print(f"  {spread:15s}: {spread_data['over_hit'].mean()*100:.1f}% ({len(spread_data):,} games)")
```

```python
# =============================================================================
# 9. EDGE DETECTION: Find the best opportunities from the matrix
# =============================================================================

print("Finding edges from the matrix...")
print("="*80)

# Calculate all combinations with sufficient sample size (min 50 games)
results = []
for tier in df_clean['line_tier'].unique():
    for spread in df_clean['spread_bin'].unique():
        subset = df_clean[(df_clean['line_tier'] == tier) & (df_clean['spread_bin'] == spread)]
        
        if len(subset) >= 50:  # Only look at combinations with 50+ games
            over_rate = subset['over_hit'].mean() * 100
            under_rate = subset['under_hit'].mean() * 100
            avg_line = subset['points_line'].mean()
            avg_pts = subset['PTS'].mean()
            avg_diff = subset['points_diff'].mean()
            
            # Calculate edge vs baseline (48.8% from earlier)
            baseline = 48.8
            over_edge = over_rate - baseline
            under_edge = under_rate - baseline
            
            results.append({
                'line_tier': tier,
                'spread_bin': spread,
                'games': len(subset),
                'over_rate': over_rate,
                'under_rate': under_rate,
                'over_edge': over_edge,
                'under_edge': under_edge,
                'avg_line': avg_line,
                'avg_pts': avg_pts,
                'avg_diff': avg_diff
            })

df_edges = pd.DataFrame(results)

# Find best over opportunities (positive edge)
print("\n🔥 TOP 10 OVER OPPORTUNITIES (highest edge vs baseline)")
print("-" * 80)
top_overs = df_edges.nlargest(10, 'over_edge')
print(top_overs[['line_tier', 'spread_bin', 'games', 'over_rate', 'over_edge', 'avg_diff']].to_string(index=False))

# Find best under opportunities
print("\n\n❄️  TOP 10 UNDER OPPORTUNITIES (highest edge vs baseline)")
print("-" * 80)
top_unders = df_edges.nlargest(10, 'under_edge')
print(top_unders[['line_tier', 'spread_bin', 'games', 'under_rate', 'under_edge', 'avg_diff']].to_string(index=False))

# Summary insights
print("\n\n💡 KEY INSIGHTS")
print("-" * 80)

# Best overall combo
best_over = df_edges.loc[df_edges['over_edge'].idxmax()]
print(f"Strongest OVER edge: {best_over['line_tier']} + {best_over['spread_bin']}")
print(f"  → {best_over['over_rate']:.1f}% hit rate (+{best_over['over_edge']:.1f}% edge) | {int(best_over['games'])} games")

best_under = df_edges.loc[df_edges['under_edge'].idxmax()]
print(f"\nStrongest UNDER edge: {best_under['line_tier']} + {best_under['spread_bin']}")
print(f"  → {best_under['under_rate']:.1f}% hit rate (+{best_under['under_edge']:.1f}% edge) | {int(best_under['games'])} games")

# Check for sample size issues
weak_samples = df_edges[df_edges['games'] < 100]
if len(weak_samples) > 0:
    print(f"\n⚠️  {len(weak_samples)} combinations have 50-100 games (be cautious of small sample size)")
```

```python
# =============================================================================
# 10. DRILL DOWN: Which players drive each edge?
# =============================================================================

print("Player breakdown for top edges...")
print("="*80)

def analyze_edge_drivers(tier, spread_bin, top_n=15):
    """Show which players are in a specific tier+spread combo"""
    subset = df_clean[(df_clean['line_tier'] == tier) & (df_clean['spread_bin'] == spread_bin)]
    
    if len(subset) == 0:
        print(f"No data for {tier} + {spread_bin}")
        return
    
    player_stats = subset.groupby('PLAYER_NAME').agg({
        'over_hit': ['sum', 'count', 'mean'],
        'PTS': 'mean',
        'points_line': 'mean',
        'points_diff': 'mean',
        'MIN': 'mean'
    }).reset_index()
    
    player_stats.columns = ['PLAYER_NAME', 'overs_hit', 'games', 'over_rate', 'avg_pts', 'avg_line', 'avg_diff', 'avg_min']
    player_stats = player_stats[player_stats['games'] >= 3]  # Min 3 games
    player_stats = player_stats.sort_values('games', ascending=False).head(top_n)
    
    print(f"\n{tier} + {spread_bin} | Total: {len(subset)} games, {subset['over_hit'].mean()*100:.1f}% overs")
    print("-" * 80)
    print(player_stats[['PLAYER_NAME', 'games', 'over_rate', 'avg_pts', 'avg_line', 'avg_diff']].to_string(index=False))

# Analyze the top over edge
if len(df_edges) > 0:
    best_over = df_edges.loc[df_edges['over_edge'].idxmax()]
    print("\n🔥 BEST OVER EDGE BREAKDOWN:")
    analyze_edge_drivers(best_over['line_tier'], best_over['spread_bin'])
    
    best_under = df_edges.loc[df_edges['under_edge'].idxmax()]
    print("\n\n❄️  BEST UNDER EDGE BREAKDOWN:")
    analyze_edge_drivers(best_under['line_tier'], best_under['spread_bin'])

# Export clean dataset for further analysis
print("\n\n" + "="*80)
print("✅ Analysis complete! Key variables created:")
print("  • df_clean = filtered dataset (no early exits)")
print("  • df_edges = all tier+spread combinations with 50+ games")
print("  • line_tier, spread_bin = binned categories")
print("\nUse analyze_edge_drivers(tier, spread_bin) to drill into any combo!")
```

```python
# =============================================================================
# 4. CUTS BY CONTEXT - Home/Away, Favorite/Dog
# =============================================================================

print("Over Hit Rate by Context:")
print("="*80)

# Home vs Away
home_over_pct = df_bets[df_bets['is_home'] == True]['over_hit'].mean() * 100
away_over_pct = df_bets[df_bets['is_home'] == False]['over_hit'].mean() * 100
print(f"\nHome games: {home_over_pct:.1f}% overs ({df_bets[df_bets['is_home'] == True]['over_hit'].sum()} / {(df_bets['is_home'] == True).sum()})")
print(f"Away games: {away_over_pct:.1f}% overs ({df_bets[df_bets['is_home'] == False]['over_hit'].sum()} / {(df_bets['is_home'] == False).sum()})")
print(f"Difference: {home_over_pct - away_over_pct:+.1f}%")

# Favorite vs Dog
fav_over_pct = df_bets[df_bets['is_favorite'] == True]['over_hit'].mean() * 100
dog_over_pct = df_bets[df_bets['is_favorite'] == False]['over_hit'].mean() * 100
print(f"\nFavorite: {fav_over_pct:.1f}% overs ({df_bets[df_bets['is_favorite'] == True]['over_hit'].sum()} / {(df_bets['is_favorite'] == True).sum()})")
print(f"Dog:      {dog_over_pct:.1f}% overs ({df_bets[df_bets['is_favorite'] == False]['over_hit'].sum()} / {(df_bets['is_favorite'] == False).sum()})")
print(f"Difference: {fav_over_pct - dog_over_pct:+.1f}%")

# Combo: Home Favorite vs Away Dog
home_fav = df_bets[(df_bets['is_home'] == True) & (df_bets['is_favorite'] == True)]
away_dog = df_bets[(df_bets['is_home'] == False) & (df_bets['is_favorite'] == False)]
print(f"\nHome Favorite: {home_fav['over_hit'].mean()*100:.1f}% overs ({len(home_fav)} games)")
print(f"Away Dog:      {away_dog['over_hit'].mean()*100:.1f}% overs ({len(away_dog)} games)")
```
